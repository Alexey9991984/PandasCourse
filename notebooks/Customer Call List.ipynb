```python
import pandas as pd
import numpy as np 
```

```python
df = pd.read_excel(r'C:\Jupyter_notebook\PandasNumPyCourse\data\Customer Call List.xlsx')
df.info() # checking the data types of the columns and if there are any missing values  
# проверка типов данных в колонках и наличия пропущенных значений
```

```python
df.head(10) # checking the first 10 rows of the data frame to understand the structure of the data
# проверка первых 10 строк датафрейма для понимания структуры данных
```

```python
df = df.drop_duplicates() # removing duplicate rows from the data frame
# удаление дублирующихся строк из датафрейма
```

```python
df['Last_Name'] = df['Last_Name'].str.strip("123._/") # removing unwanted characters from the Last_Name column
# удаление нежелательных символов из колонки Last_Name
```

```python
# Оставляем только цифры и буквы (убираем спецсимволы)
df['Phone_Number'] = df['Phone_Number'].str.replace('[^a-zA-Z0-9]', '', regex=True)
# удаление всех символов, кроме букв и цифр, из колонки Phone_Number
```

```python
df.head()
```

```python
df.info() # checking the data types of the columns and if there are any missing values after cleaning the data
# проверка типов данных в колонках и наличия пропущенных значений после очистки данных
```

```python
df = df.drop(columns=["Not_Useful_Column"]) 
# dropping the Not_Useful_Column from the data frame
# удаление колонки Not_Useful_Column из датафрейма
```

```python
df["Phone_Number"] = df["Phone_Number"].apply(lambda x: str(x))
# converting the Phone_Number column to string data type
# преобразование колонки Phone_Number в строковый тип данных
```

```python
df["Phone_Number"] = df["Phone_Number"].apply(lambda x: x[0:3] + '-' + x[3:6] + '-' + x[6:10])
# formatting the Phone_Number column to the format XXX-XXX-XXXX
# форматирование колонки Phone_Number в формат XXX-XXX-XXXX
```
