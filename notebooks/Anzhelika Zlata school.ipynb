```python
import pandas as pd
```

```python

# 👧 Anzhelika
anzhelika = {
    "history": [5, 2, 6, 6, 4, 5],
    "informatyka": [6, 4, 5, 5, 5, 6],
    "jezyk_angielski": [6, 5, 2, 5, 4, 6, 5, 6],
    "jezyk_niemiecki": [5, 3, 5, 2],
    "jezyk_polski": [3, 4, 6, 4, 4, 5, 1, 4],
    "matematyka": [3, 1, 3, 5],
    "muzyka": [4, 5],
    "plastyka": [6, 5, 5, 6, 6],
    "przyroda": [5, 6, 6, 5, 3, 3],
    "technika": [5, 5, 4, 5],
    "wf": [5, 5, 5]
}

# 👧 Zlata
zlata = {
    "history": [5, 6, 4, 4, 6, 6],
    "informatyka": [6, 5, 5, 5, 5, 5],
    "jezyk_angielski": [6, 4, 5, 6, 6, 5, 6],
    "jezyk_niemiecki": [5, 5, 5, 5, 4],
    "jezyk_polski": [4, 4, 6, 6, 5, 5, 4, 4],
    "matematyka": [4, 6, 3, 5, 5],
    "muzyka": [6, 5, 5],
    "plastyka": [6, 6, 5, 6, 6],
    "przyroda": [6, 6, 6, 4, 4],
    "technika": [5, 6, 5, 2, 5],
    "wf": [5, 5, 5]
}

# 📊 среднее
mean_a = {k: pd.Series(v).mean() for k, v in anzhelika.items()}
mean_z = {k: pd.Series(v).mean() for k, v in zlata.items()}

# 📊 количество оценок
count_a = {k: len(v) for k, v in anzhelika.items()}
count_z = {k: len(v) for k, v in zlata.items()}

# 📊 объединяем
df = pd.DataFrame({
    "Anzhelika_mean": mean_a,
    "Zlata_mean": mean_z,
    "Anzhelika_count": count_a,
    "Zlata_count": count_z
})

# 📊 разница среднего
df["Разница"] = df["Zlata_mean"] - df["Anzhelika_mean"]

print(df.round(2))

# 📊 все оценки в один список
all_a = [g for grades in anzhelika.values() for g in grades]
all_z = [g for grades in zlata.values() for g in grades]

# 📊 общее среднее
overall_a = pd.Series(all_a).mean()
overall_z = pd.Series(all_z).mean()

# 📊 общее количество оценок
count_a_total = len(all_a)
count_z_total = len(all_z)

print("\nОбщее среднее и количество оценок:")
print(f"Anzhelika: среднее = {round(overall_a, 2)}, оценок = {count_a_total}")
print(f"Zlata:     среднее = {round(overall_z, 2)}, оценок = {count_z_total}")
```
